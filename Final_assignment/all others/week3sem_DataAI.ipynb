```python
import matplotlib.pyplot as plt

# Create a histogram of passenger ages
plt.figure(figsize=(8, 6))
plt.hist(df['Age'].dropna(), bins=20, edgecolor='black') # dropna() to handle missing ages
plt.title('Distribution of Passenger Ages')
plt.xlabel('Age')
plt.ylabel('Frequency')
plt.grid(axis='y', alpha=0.75)
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Calculate the survival rate by passenger class
survival_by_pclass = df.groupby('Pclass')['Survived'].value_counts(normalize=True).unstack()

# Plot a stacked bar chart of survival rate by passenger class
survival_by_pclass.plot(kind='bar', stacked=True, figsize=(8, 6))
plt.title('Survival Rate by Passenger Class')
plt.xlabel('Passenger Class')
plt.ylabel('Proportion')
plt.xticks(rotation=0)
plt.legend(['Did Not Survive', 'Survived'])
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Calculate the survival rate by sex
survival_by_sex = df.groupby('Sex')['Survived'].value_counts(normalize=True).unstack()

# Plot a stacked bar chart of survival rate by sex
survival_by_sex.plot(kind='bar', stacked=True, figsize=(8, 6))
plt.title('Survival Rate by Sex')
plt.xlabel('Sex')
plt.ylabel('Proportion')
plt.xticks(rotation=0)
plt.legend(['Did Not Survive', 'Survived'])
plt.show()
```
